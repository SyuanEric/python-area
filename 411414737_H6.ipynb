```python
height = float(input("請輸入身高（公分）："))
weight = float(input("請輸入體重（公斤）："))

height = height * 0.01
bmi = weight / (height ** 2)

if bmi < 18.5:
    result = "過輕"
elif bmi < 24:
    result = "正常"
else:
    result = "過重"

print(f"你的 BMI 值為：{bmi:.2f}，判定為：{result}")
```
