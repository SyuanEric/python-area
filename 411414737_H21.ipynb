```python

import traceback
from datetime import datetime

try:
    num1 = float(input("請輸入被除數："))
    num2 = float(input("請輸入除數："))


    result = num1 / num2
    print("運算結果：", result)

except ZeroDivisionError:
    print("發生錯誤")


    error_time = datetime.now()


    with open("log_test.txt", "a", encoding="utf-8") as log:
        log.write("錯誤發生時間：" + str(error_time) + "\n")
        log.write("錯誤訊息：\n")
        log.write(traceback.format_exc())
        log.write("\n------------------------\n")

finally:
    print("程式結束")
```
