```python
import random
for day in range(1, 4):
    lucky_number = random.randint(1, 50)
    user_number = int(input(f"今天第{day}天,請輸入1-50數字"))
    print(f"今天的幸運數字是{lucky_number}")
    if user_number == lucky_number:
        print("恭喜你中獎了！")
        break
    else:
        print("你沒中獎，明天再來")

else:
    print("很遺憾3天都沒中獎，下次再試試吧！")
```
