```python

weapons = ["木棒", "手槍", "霰彈槍", "狙擊槍", "雷射劍"]
attacks = [10, 30, 50, 80, 100]

for i in range(len(weapons)):
    print(i + 1, weapons[i])

while True:
    num = int(input("請輸入武器編號（0 結束）："))

    if num == 0:
        break
    elif 1 <= num <= len(weapons):
        print(weapons[num - 1], "的攻擊力是", attacks[num - 1])
    else:
        print("找不到武器")

average = sum(attacks) / len(attacks)
print("平均攻擊力：", average)
```
