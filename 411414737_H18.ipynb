```python

products = {
    "蘋果": 50,
    "香蕉": 30,
    "牛奶": 60,
    "麵包": 40
}

count = {
    "蘋果": 0,
    "香蕉": 0,
    "牛奶": 0,
    "麵包": 0
}

print("=== 超市商品 ===")

for name, price in products.items():
    print(name, "：", price, "元")

while True:
    item = input("請輸入商品名稱（輸入 / 結束）：")

    if item == "/":
        break
    elif item in products:
        count[item] += 1
    else:
        print("找不到此商品！")

total = 0

print("\n=== 購買明細 ===")

for name in products:
    if count[name] > 0:
        subtotal = products[name] * count[name]
        print(name, "數量：", count[name],
              "單價：", products[name],
              "小計：", subtotal, "元")
        total += subtotal

print("總金額：", total, "元")
```
