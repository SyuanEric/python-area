```python
import random
while True:
    event = random.randint(1, 5)
    if event == 1:
        print("遇到怪物")

    elif event == 2:
        print("找到寶物")

    elif event == 3:
        print("陷阱跳過")
        continue

    elif event == 4:
        print("補充生命")

    elif event == 5:
        print("出口結束")
        break
```
