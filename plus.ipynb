```python
def c_bmi(h, w):
    bmi = w/(h / 100)** 2
    return bmi
bmi1=c_bmi(170,60)
bmi2=c_bmi(160,55)
bmi3=c_bmi(180,80)

print(f"第一位BMI{bmi1:.2f}\n第二位BMI{bmi2:.2f}\n第三位BMI{bmi3:.2f}\n")
```
