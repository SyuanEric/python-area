```python

A = []
B = []
print("請輸入A倉文具，輸入'/'結束：")
while True:
    item = input()
    if item == "/":
        break
    A.append(item)
print("請輸入B倉文具，輸入'/'結束：")
while True:
    item = input()
    if item == "/":
        break
    B.append(item)
all_items = A + B
all_items.append("膠水")

print("二倉合併清單：")
print("/".join(all_items))
```
