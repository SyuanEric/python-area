```python
order = []
print("請輸入您要點的餐點，輸入'/'結束：")
while True:
    food = input()

    if food == "/":
        break

    if food == "沙拉":
        print("今日無沙拉，已自動移除。")
        continue

    order.append(food)
order.append("冰淇淋")

print("\n完整訂單清單：")
print(", ".join(order))
```
