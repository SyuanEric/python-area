```python

def calculate(price, quantity):
    subtotal = price * quantity
    tax = subtotal * 0.05
    total = subtotal + tax
    return subtotal, tax, total

price = float(input("請輸入單價："))
quantity = int(input("請輸入數量："))

subtotal, tax, total = calculate(price, quantity)

print("小計:", subtotal)
print("稅金:", tax)
print("總金額:", total)
```
