```python
def original(p, q):
    return p * q
def discount(p, q):
    return original(p, q) * 0.9
def main():
    p = float(input("請輸入單價 p："))
    q = int(input("請輸入數量 q："))
    print("原價金額：", original(p, q), "元")
    print("9折金額：", discount(p, q), "元")

main()
```
